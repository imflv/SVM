```python
# # The Iris Setosa
from IPython.display import Image, display
url = 'http://upload.wikimedia.org/wikipedia/commons/5/56/Kosaciec_szczecinkowaty_Iris_setosa.jpg'
display(Image(url=url, width=300, height=300))
# The Iris Versicolor
url = 'http://upload.wikimedia.org/wikipedia/commons/4/41/Iris_versicolor_3.jpg'
display(Image(url=url, width=300, height=300))
# The Iris Virginica
url = 'http://upload.wikimedia.org/wikipedia/commons/9/9f/Iris_virginica.jpg'
display(Image(url=url, width=300, height=300))
import seaborn as sns

# The assignment asks to use seaborn:
try:
    iris = sns.load_dataset('iris')
except Exception:
    # Backup option in case seaborn cannot download the dataset
    from sklearn.datasets import load_iris
    data = load_iris(as_frame=True)
    iris = data.frame.rename(columns={
        'sepal length (cm)': 'sepal_length',
        'sepal width (cm)': 'sepal_width',
        'petal length (cm)': 'petal_length',
        'petal width (cm)': 'petal_width',
        'target': 'species'
    })
    iris['species'] = iris['species'].map(dict(enumerate(data.target_names)))

iris.head()
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
sns.set_style('whitegrid')

# Compatibility fix for older seaborn with newer pandas
try:
    pd.get_option('mode.use_inf_as_null')
except Exception:
    try:
        from pandas._config.config import register_option
        register_option('mode.use_inf_as_null', False)
    except Exception:
        pass
sns.pairplot(iris, hue='species', palette='Dark2', diag_kind='hist')

# Answer:
# The setosa species seems to be the most separable.
setosa = iris[iris['species'] == 'setosa']

sns.kdeplot(
    data=setosa,
    x='sepal_width',
    y='sepal_length',
    cmap='plasma',
    fill=True,
    thresh=0
)
from sklearn.model_selection import train_test_split

X = iris.drop('species', axis=1)
y = iris['species']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=101
)
from sklearn.svm import SVC

model = SVC()

model.fit(X_train, y_train)
predictions = model.predict(X_test)

from sklearn.metrics import classification_report, confusion_matrix

print(confusion_matrix(y_test, predictions))


print(classification_report(y_test, predictions))
from sklearn.model_selection import GridSearchCV

param_grid = {
    'C': [0.1, 1, 10, 100, 1000],
    'gamma': [1, 0.1, 0.01, 0.001, 0.0001]
}
grid = GridSearchCV(SVC(), param_grid, verbose=3)

grid.fit(X_train, y_train)
grid_predictions = grid.predict(X_test)

print(confusion_matrix(y_test, grid_predictions))

print(classification_report(y_test, grid_predictions))

# Answer:
# The model was already very accurate before using GridSearchCV, so there was no major improvement.
# GridSearchCV helped test different values of C and gamma to find the best parameters.
```
